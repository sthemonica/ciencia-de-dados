estrutura if else

if CONDIÇÕES:
    se ela for verdadeira, executa o que está aqui dentro
    se for falso, segue para a próxima
elif CONDIÇÕES:
    se ela for verdadeira, executa o que está aqui dentro

## Laço de repetição FOR

- ele percorre a lista toda, um dataset, excel
  
- Qual ação eu quero que seja repetida até o final dos meus dados?

```
for fruta in salada_de_frutas:

```

for ... in ... :

```
for item in lista:
    faça algo 

```

````python
vendas = [95, 200, 67, 146, 250, 297, 386]

for item in vendas:
    # item vai assumir a forma de: 
    # vendas[0], vendas[1],vendas[2], vendas[3], vendas[4], vendas[5], vendas[6]
    print(item)
    print('---')
````

````python
meta = 150
vendas = [95, 200, 67, 146, 250, 297, 386]

for venda in vendas:
    if venda < meta:
        print("Meta não batida")
    else:
        print("Meta batida")
````

````python
len(vendas)
````

````python
# aqui vamos usar um iterador

i = 0

while i < len(vendas):
    print(vendas[i])
    i += 1 # i = i + 1
    
````

````python
meta_semanal = 800
vendas = [95, 200, 67, 146, 250, 297, 386]
````

````python
dia = 0
acumulado = 0

while acumulado < meta_semanal and dia < len(vendas): 
    acumulado += vendas[dia]
    print(dia) 
    dia += 1
    
    print(acumulado)
    print('---')
````

````python
## método .get()

produtos_vendidos = ['barra', 'barra', 'barra', 'bombom', 'bombom', 'chocolate quente']
contagem = {}

for produto in produtos_vendidos:
    contagem[produto] = contagem.get(produto, 0) + 1
````
