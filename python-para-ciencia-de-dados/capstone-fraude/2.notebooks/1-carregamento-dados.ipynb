```python
documents = pd.read_json("../1.data/1.raw/documentos.json")
print(documents.shape)
documents.head()
```

```python
documents.info()
```

```python
## renomeando colunas - método .rename()
colunas = {
    "idCompra": "id_compra",
    "doc1": "entrega_doc_1",
    "doc2": "entrega_doc_2",
    "doc3": "entrega_doc_3"
}

documents = documents.rename(columns=colunas)
documents.head()
```

## Junção dos dados (como no SQL)

```python
## método concat() - concatenando os dataframes march e april
df_march_april = pd.concat([march, april], ignore_index=True)
print(df_march_april.shape)
df_march_april.head()
```

### métodos de junção como o sql - merge()

`left` 

`right`

`inner` 

`outer`

```python
## método .merge() - unindo os dataframes df_march_april e documents
df_completo = pd.merge(df_march_april, documents, how="left", on="id_compra")
df_completo.info()
```

```python
df_completo[["entrega_doc_1", "entrega_doc_2", "entrega_doc_3"]].isna().sum()
```
