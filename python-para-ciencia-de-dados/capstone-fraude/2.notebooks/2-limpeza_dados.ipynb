Possibilidades do que fazer com os dados nulos (`.isna()` ou `.isnull()` e fazendo um `.sum()`), no geral:
1. Remover os dados;
2. Preencher os numéricos com: média, mediana, 999 ou -999
3. Para preencher booleanos: se aquele dado não existe, ver se faz sentido, colocar ele como um "não"/0
4. Para preencher textos: "Desconhecido", "Não informado"


O que escolhemos fazer nos nosso dados.
- `entrega_doc_2`: nós podemos optar em colocar como `'N'`
- valores das colunas de `score`:  nesse momento, podemos optar pela mediana
- `pais`: podemos colocar como `Desconhecido`

````python
## exemplo de média e mediana com array numpy
array = np.array([1, 0, 9, 0, 5, 10, 7, 0, 10, 10])
print(f'Média: {np.mean(array)}')
print(f'Mediana: {np.median(array)}')
````

````python
df_unificado['pais'] = df_unificado['pais'].fillna('Desconhecido')
print(df_unificado['pais'].isna().sum())
print(df_unificado['pais'].value_counts())
````

````python
## com o nosso entrega_doc_2
df_unificado['entrega_doc_2'] = df_unificado['entrega_doc_2'].fillna('N')
print(df_unificado['entrega_doc_2'].isna().sum())
print(df_unificado['entrega_doc_2'].value_counts())
````

````python
df_unificado.isna().sum()
````

````python
scores_nulos = ['score_2', 'score_3', 'score_4', 'score_6', 'score_9', 'score_10']

for score in scores_nulos:
    ##mediana para preencher os valores nulos
    mediana = df_unificado[score].median()
    df_unificado[score] = df_unificado[score].fillna(mediana)
    
    print(df_unificado[score].isna().sum())
````

#### sobre a regra do nosso agrupamento

>pensamentos futuros


>países que tem mais de 10 mil dados, podemos deixar eles por eles mesmos. 
>se eles tem menos de 10 mil, podemos agrupar com outros países.

BR e AR ficam e o resto é colocado como um único agrupamento.

````python
def agrupar_paises(pais):
    if pais in ["BR", "AR"]: # a nossa lógica de deixar os países 
        return pais
    else:
        return "Outros"
````

````python
## .apply() para aplicar a função agrupar_paises em cada elemento da coluna 'pais'
# .apply() é feito em uma coluna, por isso o ponto, e dentro dos () dele passamos a função que vamos aplicar.
df_unificado['pais_agrupado'] = df_unificado['pais'].apply(agrupar_paises)
df_unificado['pais_agrupado'].value_counts()
````

### lambda - 

é uma função que é usada pontualmente e geralmente é curta

transformando a nossa função `agrupar_paises` em uma função lambda:
```
def agrupar_paises(pais):
    if pais in ["BR", "AR"]: # a nossa lógica de deixar os países 
        return pais
    else:
        return "Outros"
```

```
lambda pais: pais if pais in ["BR", "AR"] else "Outros"
```

- `def agrupar_paises(pais):` vira `lambda pais:`
- `if pais in ["BR", "AR"]: return pais` vira `pais if pais in ["BR", "AR"]`
- `else: return "Outros"` vira `else "Outros"`

````python
## criando as colunas de hora_compra e dia_semana
df_unificado['data_compra'].dtype
````

````python
dias_semana_dict = {'0': 'Segunda', 
                    '1': 'Terça', 
                    '2': 'Quarta', 
                    '3': 'Quinta', 
                    '4': 'Sexta', 
                    '5': 'Sábado',
                    '6': 'Domingo'}
````

````python
df_unificado['dia_semana'] = df_unificado['data_compra'].dt.dayofweek.astype(str).map(dias_semana_dict)
# .dt.dayofweek retorna de 0 a 6, sendo o 0 a segunda feira e 6 o domingo
````

````python
df_unificado['dia_semana'].unique()
````

````python
df_unificado['hora_compra'] = df_unificado['data_compra'].dt.strftime("%H:%M")
df_unificado['hora_compra'].head()
````
